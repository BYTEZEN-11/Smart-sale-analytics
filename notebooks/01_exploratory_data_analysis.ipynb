```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from config import CLEAN_DATA_PATH
from src.eda import perform_eda, compute_statistical_summary, compute_correlation_matrices, compute_outlier_summary

sns.set_theme(style='whitegrid', palette='deep')
plt.rcParams['figure.figsize'] = (12, 6)
print('Libraries loaded successfully.')
```

## 1. Data Ingestion & Overview

```python
df = pd.read_csv(CLEAN_DATA_PATH, parse_dates=['Order Date'])
print(f'Dataset Shape: {df.shape[0]} rows x {df.shape[1]} columns')
df.head(5)
```

## 2. Statistical Profiles & Skewness Analysis

```python
stats_df = compute_statistical_summary(df)
stats_df
```

## 3. Correlation Matrix (Pearson & Spearman)

```python
corrs = compute_correlation_matrices(df)
plt.figure(figsize=(10, 8))
sns.heatmap(corrs['pearson'], annot=True, fmt='.2f', cmap='coolwarm', center=0, square=True)
plt.title('Pearson Correlation Matrix', fontsize=14, fontweight='bold')
plt.show()
```

## 4. Outlier Diagnostics (Tukey IQR Fences)

```python
outliers_df = compute_outlier_summary(df)
outliers_df
```

## 5. Bivariate & Category Profitability Dynamics

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='Category', y='Profit Margin', palette='Set2')
plt.title('Profit Margin Distribution Across Categories', fontsize=14, fontweight='bold')
plt.ylim(-50, 100)
plt.axhline(0, color='red', linestyle='--', alpha=0.7)
plt.show()
```

## 6. Executive Findings & Key Takeaways
- **Core Revenue Driver:** Electronics dominates top-line revenue, but Furniture experiences the most volatile margins.
- **Discount Sensitivity:** Transactions with discounts above 20% suffer drastic margin contraction.
- **Regional Consistency:** West and Central regions consistently lead sales volume with healthy profit contributions.
